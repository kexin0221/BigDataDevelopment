```python
import numpy as np
from sklearn.ensemble import RandomForestClassifier
from boruta import BorutaPy
import boruta

# select data
df2 = df.drop('loan_id','loan_purpose');
df2 = df2.toPandas()

# generate separate dataframes for IVs and DV (target variable)
X = df2.drop(['label'],axis=1).values
Y = df2['label'].values

# build RandomForestClassifier, Boruta models and
# related parameter
rfc = RandomForestClassifier(n_estimators=200, n_jobs=4, class_weight='balanced', max_depth=6)
boruta_selector = BorutaPy(rfc, n_estimators='auto', verbose=2)
n_train = len(X)

# fit Boruta algorithm
boruta_selector.fit(X, Y)
```

```python
feature_df = pd.DataFrame(df2.drop(['label'], axis=1).columns.tolist(), columns=['features'])
feature_df['rank'] = boruta_selector.ranking_
feature_df = feature_df.sort_values('rank', ascending=True).reset_index(drop=True)
sns.barplot(x='rank',y='features',data=feature_df)
feature_df
```

```python
import numpy as np
import pandas as pd
import seaborn as sns
import time
import re
import os
import matplotlib.pyplot as plt
sns.set(style="ticks")

import sklearn as sk
from scipy import stats
from sklearn import preprocessing

# read the downloaded input data (marketing data)
df = pd.read_csv('loan_classification_data.csv')
```

```python
numeric_df = df._get_numeric_data()
numeric_df = numeric_df.drop(['label','is_first_loan','is_employed'], axis=1)
numeric_df.head()
```

```python
numeric_df_array = np.array(numeric_df) # converting to numpy arrays for more efficient computation

loop_c = -1
col_for_normalization = list()

for column in numeric_df_array.T:
    loop_c+=1
    x = column
    k2, p = stats.normaltest(x) 
    alpha = 0.001
    print("p = {:g}".format(p))

    # rules for printing the normality output
    if p < alpha:
        test_result = "non_normal_distr"
        col_for_normalization.append((loop_c)) # applicable if yeo-johnson is used
        
        #if min(x) > 0: # applicable if box-cox is used
            #col_for_normalization.append((loop_c)) # applicable if box-cox is used
        print("The null hypothesis can be rejected: non-normal distribution")
        
    else:
        test_result = "normal_distr"
        print("The null hypothesis cannot be rejected: normal distribution")
```

```python
columns_to_normalize = numeric_df[numeric_df.columns[col_for_normalization]]
names_col = list(columns_to_normalize)

# density plots of the features to check the normality
columns_to_normalize.plot.kde(bw_method=3)
```

```python
pt = preprocessing.PowerTransformer(method='yeo-johnson', standardize=True, copy=True)
normalized_columns = pt.fit_transform(columns_to_normalize)
normalized_columns = pd.DataFrame(normalized_columns, columns=names_col)
```

```python
def euclideanDistance(instance1,instance2,noMissingColsIndex):
    distance = 0
    for idx in noMissingColsIndex:
        distance += pow((instance1[idx] - instance2[idx]),2)
    return math.sqrt(distance)
```

```python
import operator
def getNeighbors(allInstanceNoMissing,oneInstanceWithMissing,noMissingColsIndex,k):
    distances = []
    for x in range(len(allInstanceNoMissing.columns)):
        dist = euclideanDistance(oneInstanceWithMissing, allInstanceNoMissing.iloc[x], noMissingColsIndex)
        distances.append((allInstanceNoMissing.iloc[x],dist))
    distances.sort(key=operator.itemgetter(1))
    neighbors = []
    for x in range(k):
        neighbors.append(distances[x][0])
    return neighbors
```

```python
neighbor4 = getNeighbors(df4_noMissing,df4_Missing,noMissingColsIndex,1)
df4_Missing['dependent_number'] = neighbor4[0][10]
# df4_Missing['dependent_number']
df4_Missing
```

```python
neighbor5 = getNeighbors(df5_noMissing,df5_Missing,noMissingColsIndex,1)
df5_Missing['dependent_number']=neighbor5[0][10]
df5_Missing
```
