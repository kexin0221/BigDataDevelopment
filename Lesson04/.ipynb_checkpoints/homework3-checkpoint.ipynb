```python
# Web_Visiting_Log file
from pyspark import SparkContext
sc = SparkContext()
from pyspark.sql import SQLContext
sqlc = SQLContext(sc)
```

```python
df=sqlc.read.format('com.databricks.spark.csv').options(header = 'true',inferschema = 'true')\
    .load('file:///D:/Big-Data-Science/Big-Data-Analysis-with-Python-master/Lesson04/Web_Visiting_Log.csv')
df.show(10)
```

```python
%matplotlib inline
import matplotlib.pyplot as plt
import seaborn as sns
data=df.toPandas()
sns.boxplot(x='Country',y='Age',data=data)
plt.show()
```
