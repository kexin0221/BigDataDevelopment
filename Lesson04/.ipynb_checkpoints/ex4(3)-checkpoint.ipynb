```python
noMissingColsIndex=['rate_marriage','age','children','religious']
```

```python
df_4=df.filter(df.affairs == 0)
df_4.show(5)
```

```python
df_4_Missing=df_4.where(col('yrs_married').isNull())
df_4_Missing.show()
```

```python
df_4_Missing=df_4_Missing.toPandas()
```

```python
df_4_noMissing=df_4.dropna()
df_4_noMissing.show(5)
```

```python
df_4_noMissing=df_4_noMissing.toPandas()
```

```python
def euclideanDistance(instance1,instance2,noMissingColsIndex):
    distance = 0
    for idx in noMissingColsIndex:
        distance += pow((instance1[idx] - instance2[idx]),2)
    return math.sqrt(distance)
```

```python
import operator
def getNeighbors(allInstanceNoMissing,oneInstanceWithMissing,noMissingColsIndex,k):
    distances = []
    for x in range(len(allInstanceNoMissing.columns)):
        dist = euclideanDistance(oneInstanceWithMissing, allInstanceNoMissing.iloc[x], noMissingColsIndex)
        distances.append((allInstanceNoMissing.iloc[x],dist))
    distances.sort(key=operator.itemgetter(1))
    neighbors = []
    for x in range(k):
        neighbors.append(distances[x][0])
    return neighbors
```

```python
neighbor_3=getNeighbors(df_3_noMissing,df_3_Missing,noMissingColsIndex,1)
df_3_Missing['yrs_married']=neighbor_3[0][0]
df_3_Missing
```

```python
neighbor_4=getNeighbors(df_4_noMissing,df_4_Missing,noMissingColsIndex,1)
df_4_Missing['yrs_married']=neighbor_4[0][0]
df_4_Missing
```
