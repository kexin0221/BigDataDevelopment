```python
import pandas as pd
df = pd.read_csv('bank.csv', sep=';')
data = [go.Bar(x=df.y,y=df.balance)]
iplot(data)
```

```python
iplot([go.Histogram2dContour(x=df.balance, y=df.age, contours=dict(coloring='heatmap')),
          go.Scatter(x=df.balance, y=df.age, mode='markers', marker=dict(color='red', size=8, opacity=0.3))], show_link=False)
```

```python
plot1 = go.Box(
    y=df.age,
    name = 'age of the customers',
    marker = dict(
        color = 'rgb(12, 12, 140)',
    )
)
iplot([plot1])
```
