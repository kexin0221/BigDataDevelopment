```python
from pyspark.sql import SparkSession
```

```python
spark = SparkSession.builder.appName("JokesAnalysis").getOrCreate()
```

```python
df = spark.read.option("multiLine", True).json("/jokes/stupidstuff.json")
```

```python
joke_count = df.count()
category_count = df.select("category").distinct().count()
rating_avg = df.selectExpr("avg(rating)").first()[0]
print(f"笑话总数: {joke_count}")
print(f"种类数量: {category_count}")
print(f"平均评分: {rating_avg:.2f}")
```
